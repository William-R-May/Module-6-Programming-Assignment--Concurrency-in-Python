```python
# today_file.py
# Exercise 13.1: Write the current date as a string to today.txt

from datetime import date

def write_today():
    # Get today's date
    today = date.today()
    
    # Convert to string in ISO format (YYYY-MM-DD)
    today_string = today.isoformat()
    
    # Write to file
    with open("today.txt", "w") as file:
        file.write(today_string)

if __name__ == "__main__":
    write_today()
    print("Today's date has been written to today.txt")

# read_today.py
# Exercise 13.2: Read the text file today.txt into the string today_string

def read_today():
    # Open the file in read mode
    with open("today.txt", "r") as file:
        today_string = file.read()
    return today_string

if __name__ == "__main__":
    today_string = read_today()
    print("The date read from today.txt is:", today_string)

# parse_today.py
# Exercise 13.3: Parse the date from today_string

from datetime import datetime

def parse_today():
    # Step 1: Read the string from today.txt
    with open("today.txt", "r") as file:
        today_string = file.read()

    print("Raw string from file:", today_string)

    # Step 2: Parse the string back into a date object
    parsed_date = datetime.strptime(today_string, "%Y-%m-%d").date()

    # Step 3: Show the parsed date object and its components
    print("Parsed date object:", parsed_date)
    print("Year:", parsed_date.year)
    print("Month:", parsed_date.month)
    print("Day:", parsed_date.day)

if __name__ == "__main__":
    parse_today()
```
